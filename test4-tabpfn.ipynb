```python
class TabPFNWithHead(nn.Module):
    def __init__(self, tabpfn_model, num_outputs):
        super().__init__()
        self.tabpfn = tabpfn_model
        self.head = nn.Linear(self.tabpfn.hidden_dim, num_outputs)

    def forward(self, x):
        hidden = self.tabpfn(x, return_hidden=True)  # penultimate features
        out = self.head(hidden)
        return out
```
