```python
# --- Cell 1: Imports ---
import yaml
from munch import Munch
from pathlib import Path

# Ensure yaml loader works safely
def yaml_load(path):
    with open(path, "r") as f:
        return yaml.safe_load(f)

def yaml_dump(data, path):
    with open(path, "w") as f:
        yaml.dump(data, f, sort_keys=False)
```

```python
import shutil
from pathlib import Path

def copy_and_rename_to_folders(src_ds, new_ds, n_examples):
    src_file_list = []
    dst_file_list = []
    for n in n_examples:
        src_file_list.append(f"./cfgs/t0/n{n:0>2}/t0_{src_ds}.yaml")
        src_file_list.append(f"./cfgs/t03b/n{n:0>2}/t0_{src_ds}.yaml")
        src_file_list.append(f"./cfgs/pfn/n{n:0>2}/{src_ds}.yaml")
        dst_file_list.append(f"./cfgs/t0/n{n:0>2}/t0_{new_ds}.yaml")
        dst_file_list.append(f"./cfgs/t03b/n{n:0>2}/t0_{new_ds}.yaml")
        dst_file_list.append(f"./cfgs/pfn/n{n:0>2}/{new_ds}.yaml")

    for src_file, dst_file in zip(src_file_list, dst_file_list):
        src_file = Path(src_file)        
        dest_file = Path(dst_file)
        shutil.copy2(src_file, dest_file)
        print(f"Copied to: {dest_file}")

# # Example usage
# source_file = "template.txt"
# new_filename = "config.txt"

# folders = [
#     "exp/run1",
#     "exp/run2",
#     "exp/run3"
# ]

# copy_and_rename_to_folders(source_file, new_filename, folders)
```

```python
# --- Cell 3: Load configs as Munch objects ---
configs = {}

for path in config_paths:
    data = yaml_load(path)
    configs[path] = Munch.fromDict(data)

# Inspect one example
configs[config_paths[0]]
```

```python
# --- Cell 4: Edit attributes programmatically ---
for path, cfg in configs.items():
    # cfg.hypernet.model = "bigscience/T0_3B"
    # cfg.tokenizer.model = "bigscience/T0_3B"

    # cfg.random_state = 0
    # cfg.trainer.optimizer.args.lr = 2e-4
    # cfg.trainer.optimizer.args.weight_decay = 1e-3
    # cfg.datasets.n_shots = 8
    # cfg.datasets.n_queries = 8
    # cfg.datasets.train_size = 2
    # cfg.datasets.queries_same_as_shots = True

    cfg.name = "t03b-creditg"
    cfg.note = cfg.note.replace("car", "creditg")
    cfg.datasets.list_combine_train = ["creditg"]
    cfg.datasets.list_combine_val = ["creditg"]
    cfg.datasets.list_combine_test = ["creditg"]
```

```python
# --- Cell 5: Save configs back ---
for path, cfg in configs.items():
    # Convert back to dict before dumping
    yaml_dump(cfg.toDict(), path)

print("✅ All configs updated and saved.")
```
